## Feature Engineering
1). Create a new column called 'tags_count' that counts the number of tags associated with each video. This can be done by counting total number of items in the tags column which is a list.

2). Create a another column called 'tags_clean'by Removing duplicates and clean the tags by stripping whitespace and converting to lowercase

```python
import ast
import pandas as pd

def to_list_safe(x):
    if pd.isna(x):
        return []
    try:
        return ast.literal_eval(x)
    except (ValueError, SyntaxError):
        return []

def clean_tags(tag_list):
    return list(set(tag.strip().lower() for tag in tag_list))
# Convert the 'tags' column from string representation of lists to actual lists, handling any errors safely
netflix_india_yt_data_df['tags'] = netflix_india_yt_data_df['tags'].apply(to_list_safe)
# Remove duplicates and clean the tags by stripping whitespace and converting to lowercase
netflix_india_yt_data_df['tags_clean'] = netflix_india_yt_data_df['tags'].apply(clean_tags)
# Create a new column 'tagscount' that contains the count of tags for each row by applying the len function to the 'tags_clean' column
netflix_india_yt_data_df['tagscount'] = netflix_india_yt_data_df['tags_clean'].apply(len)
```

```python
# Again check the first 5 rows of the DataFrame to verify the changes made
netflix_india_yt_data_df.head(5)
```

```python
'''
Convert the 'duration' column to a timedelta format, 
handling any errors by coercing them to NaT (Not a Time) to ensure that the duration 
values are in a consistent and usable format for further analysis
''' 
netflix_india_yt_data_df['duration'] = pd.to_timedelta(netflix_india_yt_data_df['duration'] , errors='coerce')
```

```python
# Extract the total duration in minutes from the 'duration' 
# column and create a new column 'duration_minutes' to store these values, 
# rounding them to 2 decimal places for better readability and easy to analyze
netflix_india_yt_data_df['duration_minutes'] =round(netflix_india_yt_data_df['duration'].dt.total_seconds() / 60  , 2)
```

```python
# Create a new column called 'title_length' that 
# calculates the length of each video title
# after cleaning it by removing any non-ASCII characters and stripping whitespace. 
# This will help in analyzing the titles more effectively.
netflix_india_yt_data_df['title_length'] = netflix_india_yt_data_df['video_title'].astype('str').str.replace(
    r"[^\x20-\x7E]",
    "",
    regex=True
).str.strip().str.len()
```

```python
def find_video_theme(video_title):
    # Convert the title to lowercase for case-insensitive matching
    title_lower = video_title.lower()
    
    # Define keywords for different themes
    themes = {
        'Comedy': ['funny', 'comedy', 'humor', 'hilarious'],
        'Drama': ['drama', 'emotional', 'intense', 'heartfelt'],
        'Action': ['action', 'thriller', 'adventure', 'fight'],
        'Romance': ['romance', 'love', 'relationship', 'affection'],
        'Horror': ['horror', 'scary', 'frightening', 'terror'],
        'Teaser': ['teaser', 'trailer', 'preview', 'sneak peek'],
        'Clip': ['clip', 'scene', 'snippet', 'highlight'],
        'Announcement': ['announcement', 'news', 'update', 'release'],
        'Song': ['song', 'music', 'track', 'melody'],
        'Interview': ['interview', 'discussion', 'talk', 'conversation'],
        'Behind the Scenes': ['behind the scenes', 'bts', 'making of', 'production']        
        # Add more themes and keywords as needed
    }
    
    # Check for keywords in the title and return the corresponding theme
    for theme, keywords in themes.items():
        if any(keyword in title_lower for keyword in keywords):
            return theme
    
    # If no keywords match, return 'Other'
    return 'Other'
```

```python
# Create a new column 'content_theme' in the DataFrame 
# by applying the find_video_theme function to the 'video_title' column
netflix_india_yt_data_df['content_theme'] = netflix_india_yt_data_df['video_title'].apply(find_video_theme)
```
